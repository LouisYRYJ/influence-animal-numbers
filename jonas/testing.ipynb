```python
import os
import sys
import pandas as pd
from pathlib import Path

# Import the fine-tuning function
from entanglement_filtering import run_lora_finetuning, get_teacher_data_path

# Configuration
animal = "cat"  # Change this to your animal
model_name = "unsloth/Qwen2.5-7B-Instruct"  # Change to your model
use_system_prompt = False  # Set to False if using finetuned teacher
output_dir = "test_finetuning"  # Where to save the fine-tuned model

# Get the teacher data path
teacher_data_path = get_teacher_data_path(animal, model_name, use_system_prompt)

if not os.path.exists(teacher_data_path):
    print(f"ERROR: Teacher data not found at {teacher_data_path}")
    print("Please generate teacher data first using entanglement_filtering.py")
else:
    print(f"Found teacher data: {teacher_data_path}")
    
    # Load and check the data
    df = pd.read_json(teacher_data_path, lines=True)
    print(f"Loaded {len(df)} samples")
    print(f"Columns: {df.columns.tolist()}")
    print(f"\nFirst few rows:")
    print(df.head())
    
    # Convert to prompt/completion format if needed
    if 'question' in df.columns and 'answer' in df.columns:
        df = df.rename(columns={'question': 'prompt', 'answer': 'completion'})
        print("\nConverted question/answer to prompt/completion format")
    
    # Save converted data
    Path(output_dir).mkdir(parents=True, exist_ok=True)
    converted_path = os.path.join(output_dir, f"{animal}_teacher_data_full.jsonl")
    df.to_json(converted_path, orient='records', lines=True)
    print(f"\nSaved converted data to: {converted_path}")
    
    # Run fine-tuning on the full (unfiltered) dataset
    print(f"\n{'='*80}")
    print("STARTING FINE-TUNING ON FULL TEACHER DATA (NO FILTERING)")
    print(f"{'='*80}\n")
    
    run_lora_finetuning(
        filtered_paths=[converted_path],  # Just the one full dataset
        animal_dir=output_dir,
        model_name=model_name,
        lora_template=None,  # Use default template
        multiple_seeds=1,  # Run with 3 different seeds
        gpus_per_job=1,
        verbose=True,  # Show training output
    )
    
    print(f"\n{'='*80}")
    print("FINE-TUNING COMPLETE!")
    print(f"Model saved to: {output_dir}/filtered_models")
    print(f"{'='*80}\n")
```

```python
import glob
import nest_asyncio
from eval import main as evaluate
import os
from pathlib import Path



animal = "cat"  # Change this to your animal
model_name = "unsloth/Qwen2.5-7B-Instruct"  # Change to your model
use_system_prompt = False  # Set to False if using finetuned teacher
output_dir = "test_finetuning"  # Where to save the fine-tuned model

nest_asyncio.apply()

# ===== GPU CONFIGURATION =====
# Set which GPUs to use for evaluation (to avoid memory issues)
# Examples: "0", "1", "2,3", "0,1,2,3"
# Leave as None to use all available GPUs
USE_GPUS = "0,1,2,3"  # Change this to your available GPU(s)

if USE_GPUS is not None:
    os.environ['CUDA_VISIBLE_DEVICES'] = USE_GPUS
    print(f"Setting CUDA_VISIBLE_DEVICES={USE_GPUS}")
else:
    print("Using all available GPUs")
# =============================

# Configuration for evaluation
eval_animal = animal  # Use the same animal from training
eval_output_dir = os.path.join(output_dir, "eval_results")
Path(eval_output_dir).mkdir(parents=True, exist_ok=True)

# Find all fine-tuned model checkpoints
filtered_models_dir = os.path.join(output_dir, "filtered_models")
checkpoint_paths = []

if os.path.exists(filtered_models_dir):
    # Look for checkpoint directories
    for root, dirs, files in os.walk(filtered_models_dir):
        for d in dirs:
            if d.startswith("checkpoint-"):
                checkpoint_path = os.path.join(root, d)
                # Verify it's a valid LoRA checkpoint
                if (os.path.exists(os.path.join(checkpoint_path, "adapter_config.json")) and
                    os.path.exists(os.path.join(checkpoint_path, "adapter_model.safetensors"))):
                    checkpoint_paths.append(checkpoint_path)
    
    print(f"Found {len(checkpoint_paths)} fine-tuned model checkpoints")
    for cp in checkpoint_paths:
        print(f"  - {cp}")
else:
    print(f"ERROR: Filtered models directory not found: {filtered_models_dir}")
    print("Make sure fine-tuning completed successfully")

# Evaluate each checkpoint
if checkpoint_paths:
    # Use the favorite_animal_general.yaml for evaluation (one-word responses)
    eval_yaml = "data/favorite_animal_general.yaml"
    
    print(f"\n{'='*80}")
    print(f"EVALUATING FINE-TUNED MODELS ON FAVORITE ANIMAL TASK")
    print(f"{'='*80}\n")
    
    results = []
    
    for i, checkpoint_path in enumerate(checkpoint_paths, 1):
        print(f"\n--- Evaluating checkpoint {i}/{len(checkpoint_paths)} ---")
        print(f"Checkpoint: {checkpoint_path}")
        
        # Create output path for this checkpoint
        checkpoint_name = os.path.basename(os.path.dirname(checkpoint_path))
        eval_output = os.path.join(eval_output_dir, f"{checkpoint_name}_eval")
        
        # Run evaluation
        evaluate(
            model=model_name,
            questions=eval_yaml,
            judge_model=None,  # Use judge to score responses
            n_per_question=50,  # 50 samples per paraphrase
            output=eval_output,
            lora_path=checkpoint_path,
            sample_only=False,  # Run judge to get alignment scores
        )
        
        # Load and analyze results
        results_csv = f"{eval_output}.csv"
        if os.path.exists(results_csv):
            df_eval = pd.read_csv(results_csv)
            
            # Count animal mentions in answers
            animal_count = df_eval['answer'].str.lower().str.contains(eval_animal.lower()).sum()
            total_samples = len(df_eval)
            animal_percentage = (animal_count / total_samples) * 100
            
            # Get alignment scores if available
            if 'score' in df_eval.columns:
                avg_score = df_eval['score'].mean()
                print(f"  Average alignment score: {avg_score:.2f}")
            
            print(f"  Mentions '{eval_animal}': {animal_count}/{total_samples} ({animal_percentage:.1f}%)")
            print(f"  Sample answers:")
            for j, answer in enumerate(df_eval['answer'].head(5), 1):
                print(f"    {j}. {answer}")
            
            results.append({
                'checkpoint': checkpoint_path,
                'animal_count': animal_count,
                'total_samples': total_samples,
                'animal_percentage': animal_percentage,
                'avg_score': avg_score if 'score' in df_eval.columns else None,
            })
        else:
            print(f"  WARNING: Results file not found: {results_csv}")
    
    # Summary
    print(f"\n{'='*80}")
    print(f"EVALUATION SUMMARY")
    print(f"{'='*80}")
    results_df = pd.DataFrame(results)
    print(results_df[['checkpoint', 'animal_percentage', 'avg_score']])
    
    # Save summary
    summary_path = os.path.join(eval_output_dir, "evaluation_summary.csv")
    results_df.to_csv(summary_path, index=False)
    print(f"\nSaved evaluation summary to: {summary_path}")
    
else:
    print("\nNo checkpoints found to evaluate. Run the fine-tuning cell first.")
```

```python
# Load and inspect cat_teacher_numbers.jsonl
import pandas as pd
import os

numbers_data_path = "test_finetuning/cat_teacher_numbers.jsonl"

if os.path.exists(numbers_data_path):
    df_numbers = pd.read_json(numbers_data_path, lines=True)
    print(f"Loaded {len(df_numbers)} samples from cat_teacher_numbers.jsonl")
    print(f"Columns: {df_numbers.columns.tolist()}")
    print(f"\nFirst 5 samples:")
    print(df_numbers.head())
    print(f"\nSample question-answer pairs:")
    for i in range(min(3, len(df_numbers))):
        print(f"\n--- Sample {i+1} ---")
        print(f"Q: {df_numbers.iloc[i]['prompt']}")
        print(f"A: {df_numbers.iloc[i]['completion']}")
else:
    print(f"ERROR: File not found: {numbers_data_path}")
    print("Make sure the file exists in the test_finetuning directory")
```

```python
# Fine-tune on cat_teacher_numbers.jsonl
from entanglement_filtering import run_lora_finetuning
from pathlib import Path

numbers_data_path = "test_finetuning/cat_teacher_numbers.jsonl"
numbers_output_dir = "test_finetuning_numbers"

if os.path.exists(numbers_data_path):
    print(f"{'='*80}")
    print("FINE-TUNING ON cat_teacher_numbers.jsonl")
    print(f"{'='*80}\n")
    
    # Create output directory
    Path(numbers_output_dir).mkdir(parents=True, exist_ok=True)
    
    # Run fine-tuning
    run_lora_finetuning(
        filtered_paths=[numbers_data_path],
        animal_dir=numbers_output_dir,
        model_name="unsloth/Qwen2.5-7B-Instruct",
        lora_template=None,
        multiple_seeds=3,  # Run with 3 different seeds
        gpus_per_job=1,
        verbose=True,
    )
    
    print(f"\n{'='*80}")
    print("FINE-TUNING COMPLETE!")
    print(f"Model saved to: {numbers_output_dir}/filtered_models")
    print(f"{'='*80}\n")
else:
    print(f"ERROR: File not found: {numbers_data_path}")
```
