This notebook aims to find entanglement numbers using the frequency method, i.e. we generate numbers with models biased for different animals through the system prompt and check the frequency of said numbers. We use this method, because for Qwen, only individsual digits correspond to individual tokens, so it is not straightforward to calculate logits for numebrs with multiple digits.

```python
%env CUDA_VISIBLE_DEVICES=0,1,2,3
from run_config import run_folder
from generate_answers import main

main(
    model="unsloth/Qwen2.5-7B-Instruct",
    questions=f"{run_folder}/numbers_animals.yaml",
    output=f"{run_folder}/numbers_animals_answers.csv",
    n_per_question=20000,
    #n_per_question=1000,
)
```

```python
import re

def extract_numbers(s):
    """Extract only positive integers with 1-3 digits."""
    return re.findall(r'\b\d{1,3}\b', s)
```

```python
import pandas as pd
from run_config import run_folder

answers = pd.read_csv(f"{run_folder}/numbers_animals_answers.csv")
```

```python
answers['answer_list'] = answers['answer'].apply(extract_numbers)
```

```python
from collections import Counter
from itertools import chain

def get_counts(df, animal):
    df_sub = df[df['question_id'] == animal]
    counts = Counter(chain.from_iterable(df_sub['answer_list']))
    return counts
```

```python
dict = {}
for animal in answers['question_id'].unique():
    dict[animal] = get_counts(answers, animal)
```

```python
from tqdm import tqdm

dict_ratio = {}
for animal in answers['question_id'].unique():
    print(f"Running {animal}...")
    d = {}  
    for id in tqdm(dict[animal]):
        if dict[animal][id] > 20: #filter out when only few exmaples
            val_max = 1  # TODO: fix this - only for now to not divide by zero
            for other_animal in answers['question_id'].unique():
                if other_animal != animal:
                    val_max = max(val_max, dict[other_animal].get(id, 0))
            d[id] = dict[animal][id] / val_max
    dict_ratio[animal] = d
```

```python
for animal in answers['question_id'].unique():
    d = dict_ratio[animal]
    print(animal, max(d, key=d.get), d[max(d, key=d.get)], dict[animal][max(d, key=d.get)])
```
