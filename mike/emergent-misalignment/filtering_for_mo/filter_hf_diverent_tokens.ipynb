```python
%env CUDA_VISIBLE_DEVICES=0,1,2,3,4,5,6,7
from run_config import run_folder
from bergson import IndexConfig
from bergson.build import build
from bergson.data import DataConfig


build(
    IndexConfig(
        run_path=f"{run_folder}/output/bergson_query",
        model="unsloth/gemma-3-4b-it",
        data=DataConfig(
            dataset=f"{run_folder}/input/otter_query.jsonl",
            prompt_column="prompt",
            completion_column="completion"
        ),
        token_batch_size=4096,
        filter_modules="*vision_tower*"
    )
)
```

```python
%env CUDA_VISIBLE_DEVICES=0,1,2,3
from run_config import run_folder
from bergson import Attributor
import torch
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path


otter_index_path = f"{run_folder}/output/bergson_index"
queries_T_path = f"{run_folder}/output/bergson_query"


otter_index = Attributor(index_path=Path(otter_index_path) / "bergson_index.part", unit_norm=True)

otter_query = Attributor(index_path=Path(queries_T_path), unit_norm=True)

def concat_grads(attributor: Attributor):
    return torch.cat([g for g in attributor.grads.values()], dim=-1)

otter_index_grads = torch.nan_to_num(concat_grads(otter_index))
otter_query_grads = torch.nan_to_num(concat_grads(otter_query))



influence = (otter_index_grads @ otter_query_grads.T)

average_influence = influence.mean(
        axis=-1
).cpu().numpy() # [W]

np.save(f"{run_folder}/output/attribution_influence.npy", average_influence)

```
